```python
import pickle
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import os
```

```python
def plot_agent_param_gradients(gradient_norms):
    num_agents = len(gradient_norms)
    num_steps = len(gradient_norms[0])
    num_params = len(gradient_norms[0][0])  # assuming consistent structure

    fig, axes = plt.subplots(nrows=num_agents, ncols=1, figsize=(10, 2.5 * num_agents), sharex=True)

    if num_agents == 1:
        axes = [axes]

    for i, agent_grads in enumerate(gradient_norms):
        ax = axes[i]
        agent_grads = np.array(agent_grads)  # shape: (T, k)

        for param_id in range(agent_grads.shape[1]):
            ax.plot(agent_grads[:, param_id], label=f"Param {param_id}", linewidth=1)

        # ax.plot(agent_grads.mean(axis=1), label="Mean", color='black', linestyle='--', linewidth=2)
        ax.set_ylabel("Grad Norm")
        ax.set_title(f"Agent {i}")
        ax.grid(True)
        ax.legend(loc='upper right')

    axes[-1].set_xlabel("Training Step")
    plt.tight_layout()
    plt.show()
```

```python
# load grad data
with open('dbg_stats/1000_batched_episodes_grad_stats.pkl', 'rb') as f:
    all_agents_grads = pickle.load(f)
plot_agent_param_gradients(all_agents_grads)
```
